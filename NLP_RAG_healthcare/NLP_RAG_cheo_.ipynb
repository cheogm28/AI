## Problem Statement

### Business Context

The need for quick access to comprehensive, reliable, and up-to-date medical knowledge is critical for improving patient outcomes and ensuring informed decision-making in a fast-paced environment.

Healthcare professionals often encounter information overload, struggling to sift through extensive research and data to create accurate diagnoses and treatment plans. This challenge is amplified by the need for efficiency, particularly in emergencies, where time-sensitive decisions are vital. Furthermore, access to trusted, current medical information from renowned manuals and research papers is essential for maintaining high standards of care.

To address these challenges, healthcare centers can focus on integrating systems that streamline access to medical knowledge, provide tools to support quick decision-making, and enhance efficiency. Leveraging centralized knowledge platforms and ensuring healthcare providers have continuous access to reliable resources can significantly improve patient care and operational effectiveness.

#### Downloading and Loading the model

```python
#Downloading the model from hugging face lib
model_path = hf_hub_download(
    repo_id=REPOD_ID,
    filename=FILENAME
)
# the path in the temp cache of the model to be sure it was downloded.
print(f"{model_path}")
```

```python
# creating an object to execute the model using the llama API
llama_model = Llama(
    model_path=model_path,
    n_ctx=8000,
    n_gpu_layers=-1,
    n_batch=512,
    verbose=False
)
```

### OBSERVATIONS

---
parameters used are :
* n_ctx=8000: how long is the response + prompt ( context window )
* n_gpu_layers=-1: use all GPU
* n_batch=512: number of tokesn process in parrallel
* verbose=False: do not show context in response about the model itself


#### Response

```python
# Calling the model and adding the stop to use the Llama 3 version
def response(query,max_tokens=128,temperature=0,top_p=0.95,top_k=50,repeat_penalty=1.0):
    model_output = llama_model(
      prompt=query,
      max_tokens=max_tokens,
      temperature=temperature,
      top_p=top_p,
      top_k=top_k,
      repeat_penalty=repeat_penalty,
      stop=["<|eot_id|>", "<|end_of_text|>"]
    )

    return model_output['choices'][0]['text']
```

## Question Answering using LLM with Prompt Engineering

```python
def generate_prompt(user_prompt: str, SYSTEM_INSTRUCTION = SYSTEM_INSTRUCTION ):
  new_prompt = PORMPT_TEMPLATE_FORMAT.get('sys').format(SYSTEM_INSTRUCTION) + PORMPT_TEMPLATE_FORMAT.get('user').format(user_prompt)
  return new_prompt
```

```python
# defalt resopnse parameters
response_args = {
 "max_tokens":1024,"temperature":0,"top_p":0.95,"top_k":50,"repeat_penalty":1.0
 }
```

### OBSERVATIONS

---
This is a centralized configuration and a method that is used to format the prompts in an secial fomrat that Helps in this case LLM LLama 3
* max_tokens 1024: Enables more tokens to the response so the answer can be complet
* temperature 0.1: Temperture low because we need it to be specific for a
* top_k 5: Update to get only the top 5
* top_p 0.1: Update to return only words that adds 10% of proabbility of be correct
* repeat_penalty 1.1: do not repet itself if so penalty is added to the result in this case it is neither too hig or too low

```python
USER_INSTRUCTIONS_TEMPLATE = """
### USR_CONTEXT:
Here are some documents that are relevant to the question mentioned below.
{CONTEXT}

### USR_QUESTION:
{QUESTION}
"""
```

### OBSERVATIONS


---
this are the instructions each prompt is going to have under the hood.

* the SYSTEM_ INSTRUCTIONS are the instructions the model must follow to use the context we are going to load from the DB and how the answer must be fromat, the tone of the answer, the priority of the words selected for the answer, where is the context and the user question in the prompt, and what to avoid.
* the USER_INSTRUCTIONS_TEMPLATE as it is named it is a template where we are going to add the context and the questions the user has.
* both are going to be append to create the context of the prompts we are going to pass as part of the RAG


### Response Function

```python
# function to get the response using RAG
def generate_rag_response(user_input, llama_model, k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50,repeat_penalty=1.0, ):
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=k)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    #create a Llama 3 prompt using the templates we have for systema and user
    # Combine user_prompt and system_message to create the prompt
    user_message = USER_INSTRUCTIONS_TEMPLATE.replace("{CONTEXT}", context_for_query).replace('{QUESTION}', user_input)
    new_prompt = generate_prompt(user_message)



    # Generate the response
    try:
        response = llama_model(
                  prompt=new_prompt,
                  max_tokens=max_tokens,
                  temperature=temperature,
                  top_p=top_p,
                  top_k=top_k,
                  repeat_penalty=repeat_penalty
                  )

        # Extract and print the model's response
        response = response['choices'][0]['text'].strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
```

## Output Evaluation

Let us now use the LLM-as-a-judge method to check the quality of the RAG system on two parameters - retrieval and generation. We illustrate this evaluation based on the answeres generated to the question from the previous section.

- We are using the same Mistral model for evaluation, so basically here the llm is rating itself on how well he has performed in the task.

```python
groundedness_system_message  = """
You will be presented a ###Question, ###Context used by the AI system and AI generated ###Answer.

Your task is to judge the extent to which the ###Answer is derived from ###Context.

Rate it 1 - if The ###Answer is not derived from the ###Context at all
Rate it 2 - if The ###Answer is derived from the ###Context only to a limited extent
Rate it 3 - if The ###Answer is derived from ###Context to a good extent
Rate it 4 - if The ###Answer is derived from ###Context mostly
Rate it 5 - if The ###Answer is is derived from ###Context completely

The final output should be a single overall rating in the range of 1 to 5, along with a brief explanation of the rationale for the rating.
"""
```

```python
relevance_system_message = """
You will be presented with a ###Question, the ###Context used by the AI system to generate a response, and the AI-generated ###Answer.

Your task is to judge the extent to which the ###Answer is relevant to the ###Question, considering whether it directly addresses the key aspects of the ###Question based on the provided ###Context.

Rate the relevance as follows:
- Rate 1 – The ###Answer is not relevant to the ###Question at all.
- Rate 2 – The ###Answer is only slightly relevant to the **###Question**, missing key aspects.
- Rate 3 – The ###Answer is moderately relevant, addressing some parts of the **###Question** but leaving out important details.
- Rate 4 – The ###Answer is mostly relevant, covering key aspects but with minor gaps.
- Rate 5 – The ###Answer is fully relevant, directly answering all important aspects of the **###Question** with appropriate details from the **###Context**.

The final output should be a single overall rating in the range of 1 to 5, along with a brief explanation of the rationale for the rating.
"""
```

```python
user_message_template = """
###Question
{question}

###Context
{context}

###Answer
{answer}
"""
```

### OBSERVATIONS

---

Here the prompts use to evaluate the RAG answers 2 that are a scale metric from 1 to 5 and the other a template to pass the answe, context and the response recieved so it can be evaluated

```python
def generate_ground_relevance_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):

    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=3)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    # Combine user_prompt and system_message to create the prompt

    user_message = USER_INSTRUCTIONS_TEMPLATE.replace("{CONTEXT}", context_for_query).replace('{QUESTION}', user_input)
    new_prompt = generate_prompt(user_message)

    response = llama_model(
            prompt=new_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=["<|eot_id|>", "<|end_of_text|>"],
            )

    answer =  response["choices"][0]["text"]

    # Combine user_prompt and system_message to create the prompt

    user_evaluation_body = user_message_template.format(context=context_for_query, question=user_input, answer=answer)

    groundedness_prompt = generate_prompt(user_evaluation_body,groundedness_system_message)
    relevance_prompt    = generate_prompt(user_evaluation_body,relevance_system_message)

    response_1 = llama_model(
            prompt=groundedness_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=["<|eot_id|>", "<|end_of_text|>"],
            )

    response_2 = llama_model(
            prompt=relevance_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=["<|eot_id|>", "<|end_of_text|>"],
            )

    return response_1['choices'][0]['text'],response_2['choices'][0]['text']
```

```python
def print_eval_response(response):
  print("---"*20 + "groundedness" + "---"*20)
  print(response[0])
  print("---"*20 + "---------" + "---"*20)
  print("---"*20 + "relevance" + "---"*20)
  print(response[1])
  print("---"*20 + "---------" + "---"*20)
```
